```python

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv("qoute_dataset.csv")
```

```python

df.head()
```

```python
df.shape
```

```python

quotes = df['quote']
quotes.head()
```

```python

quotes = quotes.str.lower()
```

```python
import string
translator = str.maketrans('', '', string.punctuation)
quotes = quotes.apply(lambda x: x.translate(translator))
```

```python
quotes.head()
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
vocab_size =8978
tokenizer=Tokenizer(num_words=vocab_size)
tokenizer.fit_on_texts(quotes)
```

```python
word_index=tokenizer.word_index
print(len(word_index))
list(word_index.items())[:10]
```

```python
sequences=tokenizer.texts_to_sequences(quotes)
```

```python
for i in range(3):
  print(quotes[i])
```

```python
for i in range(3):
  print(sequences[i])
```

```python
X=[]
y=[]
for seq in sequences:
  for i in range(1,len(seq)):
    X.append(seq[:i])
    y.append(seq[i])
```

```python
len(X)
```

```python
len(y)
```

```python
max_len=max(len(x) for x in X)
print(max_len)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
X_padded =pad_sequences(X,maxlen=max_len,padding="pre")
```

```python
X_padded.shape
```

```python
y=np.array(y)
```

```python
from tensorflow.keras.utils import to_categorical
y_one_hot = to_categorical(y,num_classes=vocab_size)
```

```python
y.shape
```

```python
y_one_hot.shape
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense
```

```python
embedding_dim=50
rnn_units=128
```

```python
rnn_model=Sequential()
rnn_model.add(Embedding(input_dim=vocab_size,output_dim=embedding_dim,input_length=max_len))
rnn_model.add(SimpleRNN(units=rnn_units))
rnn_model.add(Dense(units=vocab_size,activation="softmax"))
```

```python
from pandas.core.arrays import categorical
rnn_model.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
rnn_model.summary()
```

```python
lstm_model=Sequential()
lstm_model.add(Embedding(input_dim=vocab_size,output_dim=embedding_dim,input_length=max_len))
lstm_model.add(LSTM(units=rnn_units))
lstm_model.add(Dense(units=vocab_size,activation="softmax"))
```

```python
lstm_model.compile(optimizer="adam",loss="categorical_crossentropy",metrics=["accuracy"])
```

```python
lstm_model.summary()
```

```python
history = lstm_model.fit(X_padded, y_one_hot, epochs=10, batch_size=32)
```

```python
index_to_word={}
for word,index in word_index.items():
  index_to_word[index]=word
```

```python
def predictor(model,tokenizer,text,max_len):
  text =text.lower()
  seq=tokenizer.texts_to_sequences([text])[0]
  seq =pad_sequences([seq],maxlen=max_len,padding="pre")
  pred_seq=model.predict(seq,verbose=0)
  pred_word_index=np.argmax(pred_seq)
  return index_to_word[pred_word_index]
```

```python
seed_text="what are you"
next_word=predictor(lstm_model,tokenizer,seed_text,max_len)
print(seed_text,next_word)
```

```python
def generate_text(model,tokenizer,seed_text,n_words,max_len):
  for _ in range(n_words):
    next_word=predictor(model,tokenizer,seed_text,max_len)
    if next_word== "":
      break
    seed_text+=""+ next_word
  return seed_text
```

```python
def generate_text(model, tokenizer, seed_text, n_words, max_len):
  for _ in range(n_words):
    next_word = predictor(model, tokenizer, seed_text, max_len)
    if next_word == "":
      break
    seed_text += " " + next_word
  return seed_text

seed = "are you a "
generated_output = generate_text(lstm_model, tokenizer, seed, 10, max_len)
print(generated_output)
```
